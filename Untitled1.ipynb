```python
!pip install scikit-fuzzy
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
khoang_cach = ctrl.Antecedent(np.arange(0, 51, 1), 'khoang_cach')
giao_thong = ctrl.Antecedent(np.arange(0, 101, 1), 'giao_thong')
nhu_cau = ctrl.Antecedent(np.arange(0, 101, 1), 'nhu_cau')
gia_cuoc = ctrl.Consequent(np.arange(0, 101, 1), 'gia_cuoc')
khoang_cach['ngan'] = fuzz.trapmf(khoang_cach.universe, [0, 0, 2, 3])
khoang_cach['trung_binh'] = fuzz.trimf(khoang_cach.universe, [2, 5, 8])
khoang_cach['dai'] = fuzz.trapmf(khoang_cach.universe, [6, 15, 20, 50])
giao_thong['thap'] = fuzz.trimf(giao_thong.universe, [0, 0, 30])
giao_thong['trung_binh'] = fuzz.trimf(giao_thong.universe, [20, 50, 80])
giao_thong['cao'] = fuzz.trimf(giao_thong.universe, [60, 100, 100])
nhu_cau['thap'] = fuzz.trimf(nhu_cau.universe, [0, 0, 30])
nhu_cau['trung_binh'] = fuzz.trimf(nhu_cau.universe, [20, 50, 80])
nhu_cau['cao'] = fuzz.trimf(nhu_cau.universe, [60, 100, 100])
gia_cuoc['thap'] = fuzz.trimf(gia_cuoc.universe, [0, 0, 30])
gia_cuoc['trung_binh'] = fuzz.trimf(gia_cuoc.universe, [20, 50, 80])
gia_cuoc['cao'] = fuzz.trimf(gia_cuoc.universe, [70, 100, 100])
rules = [
    ctrl.Rule(khoang_cach['ngan'] & giao_thong['thap'] & nhu_cau['thap'], gia_cuoc['thap']),
    ctrl.Rule(khoang_cach['trung_binh'] & giao_thong['cao'] & nhu_cau['cao'], gia_cuoc['cao']),
    ctrl.Rule(khoang_cach['ngan'] & giao_thong['trung_binh'] & nhu_cau['cao'], gia_cuoc['trung_binh']),
    ctrl.Rule(khoang_cach['trung_binh'] & giao_thong['trung_binh'] & nhu_cau['trung_binh'], gia_cuoc['trung_binh'])
]
system_ctrl = ctrl.ControlSystem(rules)
simulation = ctrl.ControlSystemSimulation(system_ctrl)
simulation.input['khoang_cach'] = 2.5
simulation.input['giao_thong'] = 15
simulation.input['nhu_cau'] = 20
simulation.compute()
print(f"Giá cước được tính toán: {simulation.output['gia_cuoc']:.2f}/100")
gia_cuoc.view(sim=simulation)
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
danh_gia = ctrl.Antecedent(np.arange(1, 5.1, 0.1), 'danh_gia')
dung_gio = ctrl.Antecedent(np.arange(0, 101, 1), 'dung_gio')
thoi_tiet = ctrl.Antecedent(np.arange(0, 11, 1), 'thoi_tiet')
diem_thuong = ctrl.Consequent(np.arange(0, 101, 1), 'diem_thuong')
danh_gia['kem'] = fuzz.trimf(danh_gia.universe, [1, 1, 2.5])
danh_gia['trung_binh'] = fuzz.trimf(danh_gia.universe, [2.0, 3.0, 4.0])
danh_gia['tot'] = fuzz.trimf(danh_gia.universe, [3.5, 5, 5])
dung_gio['tre'] = fuzz.trapmf(dung_gio.universe, [0, 0, 40, 50])
dung_gio['dung_gio'] = fuzz.trimf(dung_gio.universe, [40, 60, 80])
dung_gio['som'] = fuzz.trapmf(dung_gio.universe, [70, 90, 100, 100])
diem_thuong['khong'] = fuzz.trimf(diem_thuong.universe, [0, 0, 20])
diem_thuong['it'] = fuzz.trimf(diem_thuong.universe, [10, 30, 50])
diem_thuong['trung_binh'] = fuzz.trimf(diem_thuong.universe, [40, 60, 80])
diem_thuong['cao'] = fuzz.trimf(diem_thuong.universe, [70, 100, 100])
rule11 = ctrl.Rule(danh_gia['tot'] & dung_gio['som'], diem_thuong['cao'])
rule12 = ctrl.Rule(danh_gia['trung_binh'] & dung_gio['dung_gio'], diem_thuong['trung_binh'])
rule13 = ctrl.Rule(danh_gia['kem'] & dung_gio['tre'], diem_thuong['khong'])
rule16 = ctrl.Rule(danh_gia['kem'] & dung_gio['tre'], diem_thuong['khong'])
reward_ctrl = ctrl.ControlSystem([rule11, rule12, rule13, rule16])
reward_sim = ctrl.ControlSystemSimulation(reward_ctrl)
reward_sim.input['danh_gia'] = 4.5
reward_sim.input['dung_gio'] = 85
reward_sim.compute()
print(f"Số điểm thưởng tài xế nhận được (thang 100): {reward_sim.output['diem_thuong']:.2f}")
diem_thuong.view(sim=reward_sim)
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
rating = ctrl.Antecedent(np.arange(1, 5.1, 0.1), 'rating')
profit = ctrl.Antecedent(np.arange(0, 101, 1), 'profit')
event = ctrl.Antecedent(np.arange(0, 11, 1), 'event')
discount = ctrl.Consequent(np.arange(0, 71, 1), 'discount')
rating['low'] = fuzz.trimf(rating.universe, [1, 1, 4])
rating['medium'] = fuzz.trimf(rating.universe, [4, 4.25, 4.5])
rating['high'] = fuzz.trimf(rating.universe, [4.5, 5, 5])
profit['low'] = fuzz.trimf(profit.universe, [0, 0, 30])
profit['medium'] = fuzz.trimf(profit.universe, [20, 50, 80])
profit['high'] = fuzz.trimf(profit.universe, [70, 100, 100])
event['none'] = fuzz.trimf(event.universe, [0, 0, 3])
event['medium'] = fuzz.trimf(event.universe, [2, 5, 8])
event['high'] = fuzz.trimf(event.universe, [7, 10, 10])
discount['very_low'] = fuzz.trimf(discount.universe, [0, 2, 5])
discount['medium'] = fuzz.trimf(discount.universe, [10, 15, 20])
discount['very_high'] = fuzz.trimf(discount.universe, [40, 55, 70])
rule1 = ctrl.Rule(rating['high'] & profit['high'], discount['very_low'])
rule2 = ctrl.Rule(rating['low'] & profit['high'], discount['very_high'])
rule3 = ctrl.Rule(event['high'], discount['very_high'])
shopee_ctrl = ctrl.ControlSystem([rule1, rule2, rule3])
shopee_sim = ctrl.ControlSystemSimulation(shopee_ctrl)
shopee_sim.input['rating'] = 4.3
shopee_sim.input['profit'] = 15
shopee_sim.input['event'] = 9
shopee_sim.compute()
print(f"Mức chiết khấu Shopee đề xuất: {shopee_sim.output['discount']:.2f}%")
discount.view(sim=shopee_sim)
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
mat_do = ctrl.Antecedent(np.arange(0, 101, 1), 'mat_do')
tai_trong = ctrl.Antecedent(np.arange(0, 101, 1), 'tai_trong')
khan_cap = ctrl.Antecedent(np.arange(0, 11, 1), 'khan_cap')
ket_hop = ctrl.Consequent(np.arange(0, 11, 1), 'ket_hop')
mat_do['thap'] = fuzz.trimf(mat_do.universe, [0, 0, 40])
mat_do['cao'] = fuzz.trimf(mat_do.universe, [60, 100, 100])
tai_trong['thap'] = fuzz.trimf(tai_trong.universe, [0, 0, 50])
tai_trong['cao'] = fuzz.trimf(tai_trong.universe, [50, 100, 100])
khan_cap['trung_binh'] = fuzz.trimf(khan_cap.universe, [3, 5, 7])
ket_hop['it'] = fuzz.trimf(ket_hop.universe, [0, 0, 4])
ket_hop['nhieu'] = fuzz.trimf(ket_hop.universe, [6, 10, 10])
rule1 = ctrl.Rule(mat_do['cao'] & tai_trong['thap'], ket_hop['nhieu'])
rule2 = ctrl.Rule(khan_cap['trung_binh'], ket_hop['it'])
logistics_ctrl = ctrl.ControlSystem([rule1, rule2])
logistics_sim = ctrl.ControlSystemSimulation(logistics_ctrl)
logistics_sim.input['mat_do'] = 85
logistics_sim.input['tai_trong'] = 20
logistics_sim.input['khan_cap'] = 5
logistics_sim.compute()
print(f"Số lượng đơn hàng đề xuất kết hợp: {logistics_sim.output['ket_hop']:.1f} đơn")
ket_hop.view(sim=logistics_sim)
```
