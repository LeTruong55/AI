```python
!pip install folium geopy numpy
import folium
import numpy as np
from geopy.distance import geodesic
import random
center = (10.77, 106.70)
vehicles = {
    f"Xe {i+1}": (
        center[0] + np.random.uniform(-0.03, 0.03),
        center[1] + np.random.uniform(-0.03, 0.03)
    ) for i in range(5)
}
customers = {
    f"KH{i+1}": (
        center[0] + np.random.uniform(-0.04, 0.04),
        center[1] + np.random.uniform(-0.04, 0.04)
    ) for i in range(20)
}
def rand_color():
    return "#{:06x}".format(random.randint(0, 0xFFFFFF))
colors = {v: rand_color() for v in vehicles}
assign = {v: [] for v in vehicles}
for cname, ccoord in customers.items():
    best_v = min(vehicles, key=lambda v: geodesic(ccoord, vehicles[v]).km)
    assign[best_v].append((cname, ccoord))
def route_length(route):
    return sum(geodesic(route[i], route[i+1]).km for i in range(len(route)-1))
def two_opt(route):
    best = route
    improved = True
    while improved:
        improved = False
        for i in range(1, len(route)-2):
            for j in range(i+1, len(route)):
                new = route[:i] + route[i:j][::-1] + route[j:]
                if route_length(new) < route_length(best):
                    best = new
                    improved = True
        route = best
    return best
m = folium.Map(location=center, zoom_start=12)
for vname, vcoord in vehicles.items():
    folium.Marker(vcoord, popup=vname,
                 icon=folium.Icon(color='black')).add_to(m)
    custs = assign[vname]
    if not custs:
        continue
    route_before = [vcoord] + [c[1] for c in custs]
    route_after = two_opt(route_before)
    folium.PolyLine(route_before, color='gray', weight=2,
                    tooltip="Trước tối ưu").add_to(m)
    folium.PolyLine(route_after, color=colors[vname], weight=4,
                    tooltip=f"Sau tối ưu - {vname}").add_to(m)
    for cname, coord in custs:
        folium.CircleMarker(coord, radius=4,
                            color=colors[vname], fill=True).add_to(m)
m
```

```python
!pip install scikit-fuzzy folium scikit-learn pandas numpy
import numpy as np
import pandas as pd
import skfuzzy as fuzz
from skfuzzy import control as ctrl
from sklearn.ensemble import RandomForestRegressor
import folium
from folium.plugins import HeatMap
import matplotlib.pyplot as plt
def solve_23_10():
    lưu_lượng = ctrl.Antecedent(np.arange(0, 101, 1), 'lưu_lượng')
    tốc_độ = ctrl.Antecedent(np.arange(0, 61, 1), 'tốc_độ')
    nguy_cơ = ctrl.Consequent(np.arange(0, 101, 1), 'nguy_cơ')
    lưu_lượng.automf(3, names=['thấp', 'trung_bình', 'cao'])
    tốc_độ.automf(3, names=['chậm', 'vừa', 'nhanh'])
    nguy_cơ['thấp'] = fuzz.trimf(nguy_cơ.universe, [0, 0, 50])
    nguy_cơ['trung_bình'] = fuzz.trimf(nguy_cơ.universe, [20, 50, 80])
    nguy_cơ['cao'] = fuzz.trimf(nguy_cơ.universe, [50, 100, 100])
    r1 = ctrl.Rule(lưu_lượng['cao'] & tốc_độ['chậm'], nguy_cơ['cao'])
    r2 = ctrl.Rule(lưu_lượng['trung_bình'], nguy_cơ['trung_bình'])
    r3 = ctrl.Rule(lưu_lượng['thấp'] | tốc_độ['nhanh'], nguy_cơ['thấp'])
    traffic_ctrl = ctrl.ControlSystem([r1, r2, r3])
    sim = ctrl.ControlSystemSimulation(traffic_ctrl)
    return sim
def solve_23_11():
    data = {
        'hour': np.random.randint(0, 24, 500),
        'is_weekend': np.random.randint(0, 2, 500),
        'area_id': np.random.randint(1, 6, 500)
    }
    df = pd.DataFrame(data)
    df['demand'] = df.apply(lambda r: 50 + np.random.randint(20, 50)
                           if (7<=r['hour']<=9 or 17<=r['hour']<=19)
                           else 10 + np.random.randint(0, 20), axis=1)
    X = df[['hour', 'is_weekend', 'area_id']]
    y = df['demand']
    model = RandomForestRegressor(n_estimators=50).fit(X, y)
    return model
fuzzy_sim = solve_23_10()
ml_model = solve_23_11()
locations = [
    {"name": "Bến Thành", "coords": [10.7719, 106.6983], "vol": 95, "speed": 5, "area_id": 1},
    {"name": "Tân Sơn Nhất", "coords": [10.8185, 106.6588], "vol": 60, "speed": 25, "area_id": 2},
    {"name": "Landmark 81", "coords": [10.7951, 106.7218], "vol": 40, "speed": 40, "area_id": 3},
    {"name": "ĐH Bách Khoa", "coords": [10.7734, 106.6595], "vol": 80, "speed": 15, "area_id": 4}
]
m = folium.Map(location=[10.78, 106.69], zoom_start=13)
heat_data = []
print("--- ĐANG PHÂN TÍCH DỮ LIỆU ---")
for loc in locations:
    fuzzy_sim.input['lưu_lượng'] = loc['vol']
    fuzzy_sim.input['tốc_độ'] = loc['speed']
    fuzzy_sim.compute()
    risk = fuzzy_sim.output['nguy_cơ']
    demand_pred = ml_model.predict([[18, 0, loc['area_id']]])[0]
    heat_data.append([loc['coords'][0], loc['coords'][1], risk])
    color = 'red' if risk > 65 else 'green'
    popup_info = (f"<b>{loc['name']}</b><br>"
                  f"Nguy cơ tắc nghẽn: {risk:.1f}%<br>"
                  f"Dự báo nhu cầu đặt xe: {demand_pred:.0f} đơn/h<br>"
                  f"Đề xuất: {'Chuyển hướng đường tránh' if risk > 65 else 'Lộ trình ổn định'}")
    folium.Marker(
        location=loc['coords'],
        popup=folium.Popup(popup_info, max_width=250),
        icon=folium.Icon(color=color, icon='info-sign')
    ).add_to(m)
HeatMap(heat_data).add_to(m)
print("Xong! Hãy chạy dòng cuối cùng 'm' để xem bản đồ.")
m
```

```python
!pip install folium pandas numpy
import folium
import pandas as pd
import numpy as np
from scipy.spatial import distance
depots = {
'Kho Trung Tâm': [10.7769, 106.7009],
'Kho Miền Đông': [10.8185, 106.7065]
}
customers = [
[10.7626, 106.6823], [10.7834, 106.6642], [10.8012, 106.6512],
[10.7543, 106.6667], [10.8231, 106.6297], [10.7950, 106.6847]
]
def find_route(start_coord, points):
    route = [start_coord]
    unvisited = points.copy()
    total_dist = 0
    current_pos = start_coord
    while unvisited:
        dists = [distance.euclidean(current_pos, p) for p in unvisited]
        closest_idx = np.argmin(dists)
        closest_point = unvisited.pop(closest_idx)
        route.append(closest_point)
        total_dist += dists[closest_idx]
        current_pos = closest_point
    route.append(start_coord)
    return route, total_dist
m = folium.Map(location=[10.78, 106.68], zoom_start=13)
colors = ['red', 'blue']
for i, (name, coord) in enumerate(depots.items()):
    assigned_customers = customers[i*3 : (i+1)*3]
    route, dist = find_route(coord, assigned_customers)
    for p in assigned_customers:
        folium.Marker(location=p, icon=folium.Icon(color='gray', icon='shopping-cart')).add_to(m)
    folium.Marker(location=coord, popup=name, icon=folium.Icon(color='green', icon='home')).add_to(m)
    folium.PolyLine(route, color=colors[i], weight=4, opacity=0.7, popup=f"Tuyến {name}").add_to(m)
    print(f"Đã tối ưu tuyến cho {name}: Tổng quãng đường đơn vị ~ {dist:.4f}")
m
```

```python
import folium
import time
import pandas as pd
import numpy as np
from IPython.display import display, clear_output
from folium import plugins
locations = {
'Trung tâm Q1': [10.7769, 106.7009],
'Sân bay TSN': [10.8185, 106.6588],
'Thủ Đức': [10.8491, 106.7731],
'Quận 7': [10.7289, 106.7082],
'Quận 12': [10.8571, 106.6416],
'Bình Tân': [10.7621, 106.6025]
}
loc_names = list(locations.keys())
np.random.seed(42)
num_vehicles = 10
vehicles = []
colors = ['red', 'blue', 'green', 'purple', 'orange', 'darkred', 'lightred', 'beige', 'darkblue', 'darkgreen']
for i in range(num_vehicles):
    start_node = loc_names[np.random.randint(0, len(loc_names))]
    target_node = loc_names[np.random.randint(0, len(loc_names))]
    while target_node == start_node:
        target_node = loc_names[np.random.randint(0, len(loc_names))]
    vehicles.append({
    'id': f'XE-{i+1:02d}',
    'pos': np.array(locations[start_node]),
    'target': np.array(locations[target_node]),
    'color': colors[i],
    'speed': np.random.uniform(0.005, 0.015)
    })
def run_fleet_simulation(steps=10):
    for step in range(steps):
        m = folium.Map(location=[10.78, 106.69], zoom_start=12, tiles='Cartodb Positron')
        for name, coord in locations.items():
            folium.CircleMarker(location=coord, radius=5, color='black', fill=True).add_to(m)
        for v in vehicles:
            direction = v['target'] - v['pos']
            dist = np.linalg.norm(direction)
            if dist > 0.005:
                move = (direction / dist) * v['speed']
                v['pos'] += move
                status = "Đang giao hàng"
            else:
                status = "Đã hoàn thành/Chờ lệnh"
            folium.Marker(
            location=v['pos'].tolist(),
            popup=f"<b>{v['id']}</b><br>Trạng thái: {status}",
            tooltip=v['id'],
            icon=folium.Icon(color=v['color'], icon='car', prefix='fa')
            ).add_to(m)
            folium.PolyLine([v['pos'], v['target']], color=v['color'], weight=1, opacity=0.3, dash_array='5').add_to(m)
        clear_output(wait=True)
        print(f"📡 HỆ THỐNG ĐIỀU PHỐI XE THỜI GIAN THỰC")
        print(f"⏱️ Bước thời gian: T + {step} | Đội xe: {num_vehicles} chiếc")
        display(m)
        time.sleep(1)
run_fleet_simulation(steps=15)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
import folium
from folium import plugins
from datetime import datetime, timedelta

np.random.seed(42)
n_data = 2000
train_data = {
    'distance': np.random.uniform(5, 60, n_data),
    'vehicle_type': np.random.choice([0, 1, 2], n_data), # 0: Van, 1: Truck, 2: Container
    'weight_ton': np.random.uniform(0.5, 20, n_data),
    'traffic_level': np.random.uniform(0, 1, n_data),
    'stops': np.random.randint(1, 4, n_data)
}
df = pd.DataFrame(train_data)

def calc_actual_time(row):
    speeds = [45, 35, 25]
    t = (row['distance'] / speeds[int(row['vehicle_type'])]) * 60
    delay = 1 + (row['traffic_level'] * 0.7) + (row['weight_ton'] * 0.03)
    return (t * delay) + (row['stops'] * 20) + np.random.normal(0, 3)

df['travel_time'] = df.apply(calc_actual_time, axis=1)
X = df[['distance', 'vehicle_type', 'weight_ton', 'traffic_level', 'stops']]
y = df['travel_time']
model_ai = GradientBoostingRegressor(n_estimators=100).fit(X, y)

hubs = {
    'Kho Tổng Q12': [10.8571, 106.6416],
    'Trung tâm Thủ Đức': [10.8491, 106.7731]
}

delivery_points = {
    'KCN Tân Tạo': [10.7621, 106.6025],
    'Cảng Cát Lái': [10.7635, 106.7758],
    'Khu Công Nghệ Cao': [10.8444, 106.8123]
}

m = folium.Map(location=[10.78, 106.70], zoom_start=11, tiles='OpenStreetMap')

def add_logistics_route(start, end, v_type, weight, traffic, stops, color):
    dist = np.sqrt((start[0]-end[0])**2 + (start[1]-end[1])**2) * 111
    pred_min = model_ai.predict([[dist, v_type, weight, traffic, stops]])[0]
    arrival = (datetime.now() + timedelta(minutes=pred_min)).strftime("%H:%M")
    folium.PolyLine([start, end], color=color, weight=5, opacity=0.8,
                    tooltip=f"Tuyến vận tải: {dist:.1f}km").add_to(m)
    v_name = ['Xe Van', 'Xe Tải', 'Container'][v_type]
    info = f"""
    <div style='width:180px; font-family:Arial; font-size:12px;'>
    <b style='color:{color}; font-size:14px;'>BÁO CÁO LOGISTICS</b><br>
    <hr style='margin:5px 0;'>
    <b>Phương tiện:</b> {v_name}<br>
    <b>Tải trọng:</b> {weight} Tấn<br>
    <b>Thời gian dự kiến:</b> {pred_min:.0f} phút<br>
    <b style='color:#d35400;'>Giờ cập bến: {arrival}</b>
    </div>
    """
    folium.Marker(location=end, popup=folium.Popup(info, max_width=200),
                  icon=folium.Icon(color='blue', icon='truck', prefix='fa')).add_to(m)

add_logistics_route(hubs['Kho Tổng Q12'], delivery_points['KCN Tân Tạo'], 1, 5.0, 0.3, 1, '#2980b9')
add_logistics_route(hubs['Kho Tổng Q12'], delivery_points['Khu Công Nghệ Cao'], 2, 18.0, 0.6, 2, '#c0392b')
add_logistics_route(hubs['Trung tâm Thủ Đức'], delivery_points['Cảng Cát Lái'], 2, 22.0, 0.8, 1, '#16a085')

for name, coord in hubs.items():
    folium.Marker(location=coord, popup=name,
                  icon=folium.Icon(color='red', icon='home', prefix='fa')).add_to(m)

folium.LayerControl().add_to(m)
plugins.Fullscreen().add_to(m)
plugins.MeasureControl(position='topleft').add_to(m)
plugins.MiniMap(toggle_display=True).add_to(m)

print("--- DASHBOARD LOGISTICS (NỀN SÁNG) ---")
m
```
